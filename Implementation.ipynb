```python
# Install required libraries for emotion detection and video processing
!pip install deepface opencv-python tf-keras > /dev/null
print("Libraries installed.")

# Import necessary libraries
import cv2  # For video processing
import numpy as np  # For numerical operations
from deepface import DeepFace  # For emotion recognition
from google.colab import files  # To upload/download files in Colab

# Ask user to upload a video
print("\n--- Please upload a SHORT video (mp4/avi) ---")
print("Note: Keep it under 10 seconds for faster processing.")
uploaded = files.upload()

# Get uploaded video path
input_video_path = next(iter(uploaded))

# Output video file name
output_video_path = "emotion_output.mp4"

# Open the video file
cap = cv2.VideoCapture(input_video_path)

# Get video properties (width, height, fps, total frames)
width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps    = cap.get(cv2.CAP_PROP_FPS)
frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

# Define video writer to save processed video
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video_path, fourcc, fps, (width, height))

print(f"Processing {frame_count} frames... this may take a minute.")

frame_num = 0  # Counter for frames

# Loop through video frames
while cap.isOpened():
    ret, frame = cap.read()  # Read frame
    if not ret:
        break  # Stop if video ends

    frame_num += 1

    # Analyze emotion every 5 frames (to improve speed)
    if frame_num % 5 == 0 or frame_num == 1:
        try:
            results = DeepFace.analyze(
                frame,
                actions=['emotion'],  # Detect emotions only
                enforce_detection=False  # Avoid crash if no face detected
            )

            # Ensure results is a list
            if not isinstance(results, list):
                results = [results]

            current_faces = results  # Store detected faces

        except:
            current_faces = []  # If error, skip detection

    # Draw results on each detected face
    for face_data in current_faces:
        try:
            # Get face coordinates
            x = face_data['region']['x']
            y = face_data['region']['y']
            w = face_data['region']['w']
            h = face_data['region']['h']

            # Get dominant emotion
            emotion = face_data['dominant_emotion']

            # Draw rectangle around face
            cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

            # Draw label background
            cv2.rectangle(frame, (x, y-30), (x+w, y), (0, 255, 0), -1)

            # Put emotion text
            cv2.putText(
                frame,
                emotion,
                (x+5, y-10),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.8,
                (255, 255, 255),
                2
            )

        except:
            pass  # Skip if any face data is invalid

    # Write processed frame to output video
    out.write(frame)

    # Print progress every 20 frames
    if frame_num % 20 == 0:
        print(f"Processed {frame_num}/{frame_count} frames...")

# Release video resources
cap.release()
out.release()

# Download processed video
print("Done! Downloading your processed video...")
files.download(output_video_path)
```

```python
# Import libraries to run JavaScript in Colab and handle image decoding
from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode

# Function to capture a photo using the webcam
def take_photo(filename='photo.jpg', quality=0.8):

  # JavaScript code to open webcam and capture image
  js = Javascript('''
    async function takePhoto(quality) {
      const div = document.createElement('div');
      const capture = document.createElement('button');
      capture.textContent = 'Capture';
      div.appendChild(capture);

      const video = document.createElement('video');
      video.style.display = 'block';
      const stream = await navigator.mediaDevices.getUserMedia({video: true});

      document.body.appendChild(div);
      div.appendChild(video);
      video.srcObject = stream;
      await video.play();

      google.colab.output.setIframeHeight(document.documentElement.scrollHeight, true);

      await new Promise((resolve) => capture.onclick = resolve);

      const canvas = document.createElement('canvas');
      canvas.width = video.videoWidth;
      canvas.height = video.videoHeight;
      canvas.getContext('2d').drawImage(video, 0, 0);
      stream.getVideoTracks()[0].stop();
      div.remove();
      return canvas.toDataURL('image/jpeg', quality);
    }
    ''')

  # Display JavaScript in Colab
  display(js)

   # Run JS function and get image data
  data = eval_js('takePhoto({})'.format(quality))

  # Decode base64 image
  binary = b64decode(data.split(',')[1])

  # Save image to file
  with open(filename, 'wb') as f:
    f.write(binary)

  return filename


print("Click 'Capture' to take a picture for your report...")
try:
  filename = take_photo()
  print('Saved to {}'.format(filename))
except Exception as err:
  print(str(err))
```

```python
# Install DeepFace library for emotion detection
!pip install deepface

# Import required libraries
import cv2  # For image processing
import matplotlib.pyplot as plt  # For displaying image
from deepface import DeepFace  # For emotion recognition

# Path to input image
image_path = "photo.jpg"

# Read the image using OpenCV
img = cv2.imread(image_path)

# Convert image from BGR (OpenCV format) to RGB (DeepFace expects RGB)
rgb_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

print("Analyzing emotion...")

try:
    # Analyze the image to detect emotion
    results = DeepFace.analyze(
        rgb_img,
        actions=['emotion'],  # Only detect emotions
        enforce_detection=True  # Force face detection (error if no face)
    )

    # Extract dominant emotion and full emotion scores
    dominant_emotion = results[0]['dominant_emotion']
    emotion_details = results[0]['emotion']

    # Get face bounding box coordinates
    x = results[0]['region']['x']
    y = results[0]['region']['y']
    w = results[0]['region']['w']
    h = results[0]['region']['h']

    # Draw rectangle around detected face
    cv2.rectangle(img, (x, y), (x + w, y + h), (0, 255, 0), 3)

    # Create label text for dominant emotion
    label = f"Emotion: {dominant_emotion.upper()}"

    # Put label above the face
    cv2.putText(
        img,
        label,
        (x, y - 10),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        (0, 255, 0),
        2,
        cv2.LINE_AA
    )

    # Convert image back to RGB for displaying with matplotlib
    annotated_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Print results in console
    print("\n--- DETECTION SUCCESSFUL ---")
    print(f"The DOMINANT EMOTION is: {dominant_emotion.upper()}")

    print("\nFull Emotion Confidence Scores (Percentage):")
    for emotion, score in emotion_details.items():
        print(f"- {emotion.capitalize()}: {score:.2f}%")

    # Display image with detected emotion
    plt.figure(figsize=(8, 8))
    plt.imshow(annotated_img)
    plt.axis('off')
    plt.title(f"Detected Emotion: {dominant_emotion.upper()}")
    plt.show()

# Handle error if no face is detected
except ValueError as e:
    print("\n--- ERROR ---")
    print(f"Face Detection Error: {e}")
    print("Please ensure your photo has a clearly visible face.")
```

```python
# Create a folder named "test_emotions" (no error if it already exists)
!mkdir -p test_emotions

# Import required libraries
from google.colab import files  # For uploading files
import shutil  # For moving files
import os  # For file system operations
import cv2  # For image processing
from deepface import DeepFace  # For emotion detection
import glob  # For retrieving file paths

# Ask user to upload test images
print(" Upload your emotion test images now...")
uploaded = files.upload()

# Move uploaded images into the test_emotions folder
for filename in uploaded.keys():
    shutil.move(filename, f"test_emotions/{filename}")

# Confirm files moved
print("\n Images moved to test_emotions/ folder.")
print(" Current files:", os.listdir("test_emotions"))

# Get all image paths (.jpg and .png)
image_paths = glob.glob("test_emotions/*.jpg") + glob.glob("test_emotions/*.png")

# Check if folder is empty
if len(image_paths) == 0:
    print(" ERROR: No images found in test_emotions/ folder!")
else:
    correct = 0  # Count correct predictions
    total = 0  # Count total images

    print(f"\n Found {len(image_paths)} images. Starting accuracy test...\n")

    # Loop through each image
    for img_path in image_paths:

        # Get image file name
        filename = os.path.basename(img_path)
        print(f" Image: {filename}")

        # Ask user to input the true emotion (manual labeling)
        true = input(" Enter TRUE emotion (happy, sad, angry, neutral, fear, disgust, surprise): ").strip().lower()

        # Read image
        img = cv2.imread(img_path)

        # Analyze emotion using DeepFace
        result = DeepFace.analyze(img, actions=['emotion'], enforce_detection=False)

        # Get predicted emotion
        predicted = result[0]['dominant_emotion']

        # Display prediction vs actual
        print(f"    Model Prediction: {predicted}")
        print(f"    Your Label:       {true}")

        # Check if prediction is correct
        if predicted.lower() == true:
            print("    Correct!\n")
            correct += 1
        else:
            print("    Incorrect\n")

        total += 1  # Increment total count

    # Calculate accuracy
    accuracy = (correct / total) * 100

    # Print final results
    print("")
    print(f" Total Images: {total}")
    print(f" Correct Predictions: {correct}")
    print(f" Model Accuracy: {accuracy:.2f}%")
    print("")
```

```python
# 1. Install necessary library
!pip install deepface gradio tf-keras opencv-python -q

import gradio as gr
from deepface import DeepFace

# 2. Simplified function for the Interface
def upload_and_classify(img):
    try:
        # Simple analysis focused only on emotion
        results = DeepFace.analyze(img, actions=['emotion'], enforce_detection=False)

        # Extract results
        if isinstance(results, list):
            res = results[0]
        else:
            res = results

        emotion = res['dominant_emotion']
        confidence = res['emotion'][emotion]

        return f"Result: {emotion.upper()} ({confidence:.2f}%)"
    except Exception as e:
        return "Error: Could not process image. Please try another one."

# 3. Building the Interface
# We use .launch(inline=True) to keep it inside the notebook
app = gr.Interface(
    fn=upload_and_classify,
    inputs=gr.Image(label="Select Image"),
    outputs=gr.Textbox(label="Emotion Result"),
    title="Emotion Recognition Interface",
    description="Upload an image to identify the facial expression."
)

# Launching inside the Colab cell
app.launch(inline=True, share=False)
```

```python
# Import libraries to run JavaScript in Colab and handle image decoding
from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode

# Function to capture a photo using the webcam
def take_photo(filename='photo.jpg', quality=0.8):

  # JavaScript code to open webcam and capture image
  js = Javascript('''
    async function takePhoto(quality) {
      const div = document.createElement('div');
      const capture = document.createElement('button');
      capture.textContent = 'Capture';
      div.appendChild(capture);

      const video = document.createElement('video');
      video.style.display = 'block';
      const stream = await navigator.mediaDevices.getUserMedia({video: true});

      document.body.appendChild(div);
      div.appendChild(video);
      video.srcObject = stream;
      await video.play();

      google.colab.output.setIframeHeight(document.documentElement.scrollHeight, true);

      await new Promise((resolve) => capture.onclick = resolve);

      const canvas = document.createElement('canvas');
      canvas.width = video.videoWidth;
      canvas.height = video.videoHeight;
      canvas.getContext('2d').drawImage(video, 0, 0);
      stream.getVideoTracks()[0].stop();
      div.remove();
      return canvas.toDataURL('image/jpeg', quality);
    }
    ''')

  # Display JavaScript in Colab
  display(js)

   # Run JS function and get image data
  data = eval_js('takePhoto({})'.format(quality))

  # Decode base64 image
  binary = b64decode(data.split(',')[1])

  # Save image to file
  with open(filename, 'wb') as f:
    f.write(binary)

  return filename


print("Click 'Capture' to take a picture for your report...")
try:
  filename = take_photo()
  print('Saved to {}'.format(filename))
except Exception as err:
  print(str(err))
```

```python
# Install DeepFace library for emotion detection
!pip install deepface

# Import required libraries
import cv2  # For image processing
import matplotlib.pyplot as plt  # For displaying image
from deepface import DeepFace  # For emotion recognition

# Path to input image
image_path = "photo.jpg"

# Read the image using OpenCV
img = cv2.imread(image_path)

# Convert image from BGR (OpenCV format) to RGB (DeepFace expects RGB)
rgb_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

print("Analyzing emotion...")

try:
    # Analyze the image to detect emotion
    results = DeepFace.analyze(
        rgb_img,
        actions=['emotion'],  # Only detect emotions
        enforce_detection=True  # Force face detection (error if no face)
    )

    # Extract dominant emotion and full emotion scores
    dominant_emotion = results[0]['dominant_emotion']
    emotion_details = results[0]['emotion']

    # Get face bounding box coordinates
    x = results[0]['region']['x']
    y = results[0]['region']['y']
    w = results[0]['region']['w']
    h = results[0]['region']['h']

    # Draw rectangle around detected face
    cv2.rectangle(img, (x, y), (x + w, y + h), (0, 255, 0), 3)

    # Create label text for dominant emotion
    label = f"Emotion: {dominant_emotion.upper()}"

    # Put label above the face
    cv2.putText(
        img,
        label,
        (x, y - 10),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        (0, 255, 0),
        2,
        cv2.LINE_AA
    )

    # Convert image back to RGB for displaying with matplotlib
    annotated_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Print results in console
    print("\n--- DETECTION SUCCESSFUL ---")
    print(f"The DOMINANT EMOTION is: {dominant_emotion.upper()}")

    print("\nFull Emotion Confidence Scores (Percentage):")
    for emotion, score in emotion_details.items():
        print(f"- {emotion.capitalize()}: {score:.2f}%")

    # Display image with detected emotion
    plt.figure(figsize=(8, 8))
    plt.imshow(annotated_img)
    plt.axis('off')
    plt.title(f"Detected Emotion: {dominant_emotion.upper()}")
    plt.show()

# Handle error if no face is detected
except ValueError as e:
    print("\n--- ERROR ---")
    print(f"Face Detection Error: {e}")
    print("Please ensure your photo has a clearly visible face.")
```

```python
# Install required libraries for emotion detection and video processing
!pip install deepface opencv-python tf-keras > /dev/null
print("Libraries installed.")

# Import necessary libraries
import cv2  # For video processing
import numpy as np  # For numerical operations
from deepface import DeepFace  # For emotion recognition
from google.colab import files  # To upload/download files in Colab

# Ask user to upload a video
print("\n--- Please upload a SHORT video (mp4/avi) ---")
print("Note: Keep it under 10 seconds for faster processing.")
uploaded = files.upload()

# Get uploaded video path
input_video_path = next(iter(uploaded))

# Output video file name
output_video_path = "emotion_output.mp4"

# Open the video file
cap = cv2.VideoCapture(input_video_path)

# Get video properties (width, height, fps, total frames)
width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps    = cap.get(cv2.CAP_PROP_FPS)
frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

# Define video writer to save processed video
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video_path, fourcc, fps, (width, height))

print(f"Processing {frame_count} frames... this may take a minute.")

frame_num = 0  # Counter for frames

# Loop through video frames
while cap.isOpened():
    ret, frame = cap.read()  # Read frame
    if not ret:
        break  # Stop if video ends

    frame_num += 1

    # Analyze emotion every 5 frames (to improve speed)
    if frame_num % 5 == 0 or frame_num == 1:
        try:
            results = DeepFace.analyze(
                frame,
                actions=['emotion'],  # Detect emotions only
                enforce_detection=False  # Avoid crash if no face detected
            )

            # Ensure results is a list
            if not isinstance(results, list):
                results = [results]

            current_faces = results  # Store detected faces

        except:
            current_faces = []  # If error, skip detection

    # Draw results on each detected face
    for face_data in current_faces:
        try:
            # Get face coordinates
            x = face_data['region']['x']
            y = face_data['region']['y']
            w = face_data['region']['w']
            h = face_data['region']['h']

            # Get dominant emotion
            emotion = face_data['dominant_emotion']

            # Draw rectangle around face
            cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

            # Draw label background
            cv2.rectangle(frame, (x, y-30), (x+w, y), (0, 255, 0), -1)

            # Put emotion text
            cv2.putText(
                frame,
                emotion,
                (x+5, y-10),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.8,
                (255, 255, 255),
                2
            )

        except:
            pass  # Skip if any face data is invalid

    # Write processed frame to output video
    out.write(frame)

    # Print progress every 20 frames
    if frame_num % 20 == 0:
        print(f"Processed {frame_num}/{frame_count} frames...")

# Release video resources
cap.release()
out.release()

# Download processed video
print("Done! Downloading your processed video...")
files.download(output_video_path)
```
